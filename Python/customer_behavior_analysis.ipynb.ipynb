## 4. Missing Value Treatment

The `Review Rating` column contains missing values. Since customer ratings can vary by product category, missing values are filled using the median rating of each category instead of using one overall average.

```python
# Check missing values before treatment
missing_before = df['Review Rating'].isnull().sum()

print("Missing values in Review Rating before treatment:", missing_before)
```

```python
# Fill missing Review Rating values using category-wise median
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(
    lambda x: x.fillna(x.median())
)
```

```python
# Verify missing values after treatment
missing_after = df['Review Rating'].isnull().sum()

print("Missing values in Review Rating after treatment:", missing_after)
```

## 5. Column Standardization

Column names are standardized into `snake_case` format to make the dataset easier to use in Python, SQL, and Power BI.

```python
# Standardize column names
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(' ', '_')
    .str.replace(r'[()]', '', regex=True)
)

# Rename purchase amount column for better readability
df = df.rename(columns={
    'purchase_amount_usd': 'purchase_amount'
})

# Display updated column names
df.columns.tolist()
```

## 7. Customer Age Segmentation

Customers are segmented into age groups to analyze purchasing behavior across different life stages.  
Instead of using equal distribution (quantiles), age groups are defined based on real-world demographic categories.

```python
# Create age groups using business logic (not quantiles)
df['age_group'] = pd.cut(
    df['age'],
    bins=[18, 25, 40, 60, 100],
    labels=['Young Adult', 'Adult', 'Middle-Aged', 'Senior']
)
# Preview result
df[['age', 'age_group']].head(10)
```

## 8. Purchase Frequency Transformation

The purchase frequency column is converted into numerical values (in days) to enable quantitative analysis.  
This helps in comparing customer engagement levels more effectively.

```python
# Convert purchase frequency to number of days
frequency_mapping = {
    'Weekly': 7,
    'Bi-Weekly': 14,
    'Fortnightly': 14,
    'Monthly': 30,
    'Quarterly': 90,
    'Every 3 Months': 90,
    'Annually': 365
}
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)

# Preview mapping
df[['frequency_of_purchases', 'purchase_frequency_days']].head(10)
```

## 9. Customer Engagement Score

A custom engagement score is created using purchase frequency and previous purchase history.  
This helps identify highly engaged customers for targeted business strategies.

```python
# Create engagement score
df['engagement_score'] = (
    df['previous_purchases'] * 0.6 +
    (1 / df['purchase_frequency_days']) * 100 * 0.4
)

# Preview
df[['previous_purchases', 'purchase_frequency_days', 'engagement_score']].head(10)
```

## 10. Redundant Feature Check

The dataset contains both `discount_applied` and `promo_code_used`.  
Before removing any column, both fields are compared to check whether they provide the same information.

```python
# Compare discount_applied and promo_code_used
discount_promo_check = pd.crosstab(
    df['discount_applied'],
    df['promo_code_used']
)

discount_promo_check
```

```python
# Check if both columns contain identical values
is_redundant = (df['discount_applied'] == df['promo_code_used']).all()

print("Are discount_applied and promo_code_used identical?", is_redundant)
```

Since both columns contain identical information, `promo_code_used` is removed to avoid duplicate information in the analysis.

```python
# Drop redundant column
df = df.drop(columns=['promo_code_used'])
```

```python
# Verify updated columns
df.columns.tolist()
```

## 11. Business-Focused Analysis

### Insight

Customers are segmented into low, medium, and high-value groups based on purchase amount.  
A smaller proportion of high-value customers contributes significantly to total revenue, indicating a potential opportunity to focus retention strategies on this segment.

```python
customer_segment = pd.qcut(
    df['purchase_amount'],
    q=3,
    labels=['Low Value', 'Medium Value', 'High Value']
)

df['customer_segment'] = customer_segment

df['customer_segment'].value_counts()
```

### Insight

Non-subscribers generate higher total revenue primarily due to a larger customer base.  
However, subscribers tend to have more consistent purchasing behavior, suggesting an opportunity to increase subscription adoption among high-value customers.

```python
subscription_analysis = df.groupby('subscription_status').agg(
    total_customers=('customer_id', 'nunique'),
    avg_purchase=('purchase_amount', 'mean'),
    total_revenue=('purchase_amount', 'sum')
).reset_index()

subscription_analysis
```

### Insight

Customers using discounts contribute significantly to overall revenue, but their average purchase amount is comparable to non-discount users.  
This suggests that discounts drive volume rather than increasing individual transaction value.

```python
discount_analysis = df.groupby('discount_applied').agg(
    total_customers=('customer_id', 'nunique'),
    avg_purchase=('purchase_amount', 'mean'),
    total_revenue=('purchase_amount', 'sum')
).reset_index()

discount_analysis
```

### Insight

Middle-aged and adult customer segments contribute the highest share of revenue.  
This indicates that marketing strategies should focus on these demographics for better returns.

```python
age_analysis = df.groupby('age_group', observed=True).agg(
    total_customers=('customer_id', 'nunique'),
    avg_purchase=('purchase_amount', 'mean'),
    total_revenue=('purchase_amount', 'sum')
).reset_index()

age_analysis
```

## Final Business Insights

- A small segment of high-value customers contributes disproportionately to revenue.
- Subscription-based customers show more consistent purchasing patterns.
- Discounts increase sales volume but do not significantly increase per-transaction value.
- Middle-aged and adult customers drive the majority of revenue.

These insights can help businesses optimize marketing strategies, customer retention efforts, and pricing decisions.
